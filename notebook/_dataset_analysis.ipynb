```python
import zipfile
from collections import Counter
from pathlib import PurePosixPath

zip_path = r"E:\PlantVillage\data\data.zip"

class_counts = Counter()

with zipfile.ZipFile(zip_path, "r") as archive:
    for name in archive.namelist():
        path = PurePosixPath(name)

        if (
            len(path.parts) >= 3
            and path.parts[0] == "raw"
            and path.parts[1] == "color"
            and path.suffix.lower() in {".jpg", ".jpeg", ".png"}
        ):
            class_name = path.parts[2]
            class_counts[class_name] += 1

print("Number of classes:", len(class_counts))
print("\nImages per class:")

for class_name, count in sorted(class_counts.items()):
    print(f"{class_name}: {count}")

print("\nTotal color images:", sum(class_counts.values()))
```

```python
import zipfile
from pathlib import Path

zip_path = Path(r"E:\PlantVillage\data\data.zip")
extract_dir = Path(r"E:\PlantVillage\data\images")

extract_dir.mkdir(parents=True, exist_ok=True)

with zipfile.ZipFile(zip_path, "r") as archive:
    color_files = [
        name for name in archive.namelist()
        if name.startswith("raw/color/")
        and not name.endswith("/")
    ]

    print("Files to extract:", len(color_files))

    for i, name in enumerate(color_files, start=1):
        target = extract_dir / Path(name).relative_to("raw/color")
        target.parent.mkdir(parents=True, exist_ok=True)

        if not target.exists():
            with archive.open(name) as source, open(target, "wb") as destination:
                destination.write(source.read())

        if i % 5000 == 0:
            print(f"Processed {i}/{len(color_files)} files")

print("Extraction complete!")
print("Images location:", extract_dir)
```

```python
from pathlib import Path
import json

metadata_dir = Path(r"E:\PlantVillage\data\metadata")

train_file = metadata_dir / "splits" / "color_train.txt"
test_file = metadata_dir / "splits" / "color_test.txt"
leaf_file = metadata_dir / "leaf_grouping" / "leaf-map.json"

with open(train_file, encoding="utf-8") as f:
    train_paths = [line.strip() for line in f if line.strip()]

with open(test_file, encoding="utf-8") as f:
    test_paths = [line.strip() for line in f if line.strip()]

with open(leaf_file, encoding="utf-8") as f:
    leaf_map = json.load(f)

print("Official training paths:", len(train_paths))
print("Official test paths:", len(test_paths))
print("Combined paths:", len(train_paths) + len(test_paths))
print("Duplicate paths across splits:", len(set(train_paths) & set(test_paths)))

print("\nLeaf-map sample:")
for key, value in list(leaf_map.items())[:5]:
    print(repr(key), "->", repr(value))
```

```python
from pathlib import Path

images_dir = Path(r"E:\PlantVillage\data\images")

def local_image_path(official_path):
    # Convert raw/color/ClassName/image.JPG to the extracted path
    parts = Path(official_path).parts
    return images_dir.joinpath(*parts[2:])

train_local = [local_image_path(p) for p in train_paths]
test_local = [local_image_path(p) for p in test_paths]

missing_train = [p for p in train_local if not p.is_file()]
missing_test = [p for p in test_local if not p.is_file()]

print("Missing training images:", len(missing_train))
print("Missing test images:", len(missing_test))

if missing_train:
    print("Example missing train path:", missing_train[0])

if missing_test:
    print("Example missing test path:", missing_test[0])
```
